```python
from ultralytics import YOLO
model = YOLO('yolo11n.pt')
results = model.train(
  data='data.yaml',
  epochs=40,
  imgsz=640,
  batch=16,
  patience=20,
  device=0,
  project='helmet',
  name='yolo11n_ft',
  optimizer='auto'
)
```

```python
# Validate (plots include confusion matrix)
model = YOLO('helmet/yolo11n_ft/weights/best.pt')
model.val(data='data.yaml', device=0, imgsz=640, plots=True)
```
